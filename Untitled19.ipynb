```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'Ename':['roshan','amar','ashwin','lohit','mohan','pramod'],
    'Department':['cs','cs','ec','ec','cs','ec'],
    'Experience':[10,20,15,14,9,8],
    'Salary':[45000,50000,30000,25000,10000,40000]
}
df=pd.DataFrame(data)
print(df)
table = pd.pivot_table(df, values='Salary', index=['Ename'], columns=['Department'], aggfunc=np.mean)
print(table)
table = pd.pivot_table(df, values='Salary', index=['Ename'], aggfunc=['sum', 'mean'])
print(table)
```
