```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_df=pd.read_csv('mtcars (9).csv')
print(cars_df)
a=plt.scatter(cars_df['mpg'],cars_df['wt'],marker='*')
print(a)
plt.xlabel('mpg')
plt.ylabel('weight')
plt.title('Relationship between the  wieght of the cars and mpg')
plt.show()
trans=cars_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel("transmission")
plt.title('barplot to check frequency of the trassmission type')
plt.show()
cars_df['mpg'].plot.box()
plt.title('Box and whisker plot of mpg')
plt.show()
minvalue=np.min(cars_df['mpg'])
q1=np.percentile(cars_df['mpg'],25)
median=np.median(cars_df['mpg'])
q3=np.percentile(cars_df['mpg'],75)
maxvalue=np.max(cars_df['mpg'])
print('minvalue:',minvalue)
print('Q1:',q1)
print('median:',median)
print('Q3:',q3)
print('maxvalue:',maxvalue)
```
