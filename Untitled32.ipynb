```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
df=pd.read_csv('Cars93 (1).csv')
print(df)
fig,ax=plt.subplots(figsize=(10,4))
ax.scatter(df['Horsepower'],df['Price'])
ax.set_xlabel('Horsepower')
ax.set_ylabel('Price')
plt.title("scatterplot")
plt.show()
df['Horsepower'].plot(kind='box')
q1=df["Horsepower"].quantile(0.25)
q3=df["Horsepower"].quantile(0.75)
iqr=q3-q1
upper_bound=q3+(1.5*iqr)
lower_bound=q1-(1.5*iqr)
Horsepower_arr=df['Horsepower']
outliers=Horsepower_arr[(Horsepower_arr<=lower_bound) | (Horsepower_arr>=upper_bound)]
print('The following are the outliers in the boxplot of Horsepower:\n',outliers)
plt.show()
df['Price'].plot(kind='hist')
upperIndex=df[df['Horsepower']>upper_bound].index
df.drop(upperIndex,inplace=True)
lowerIndex=df[df['Horsepower']<lower_bound].index
df.drop(lowerIndex,inplace=True)
print(df.info())
```
