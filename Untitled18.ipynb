```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    'number':[1,2,3,4,5],
    'Pencil':[300,350,400,500,250],
    'Textbook':[250,350,400,420,500],
    'Drawingsheet':[100,200,200,250,300],
    'Total unit':[800,1000,1320,1510,2000],
    'Profit':[8000,9500,10256,12000,18000]
}
df=pd.DataFrame(data)
print(df)
df.info()
plt.plot (df['number'],df['Profit'])
plt.xlabel("number")
plt.ylabel("total profit")
plt.title("Statistical information")
plt.show()
print("missing value")
print(df.isnull().sum())
print("sum pof totalprofit")
print(df["Profit"].sum())
print("max value of drawingsheet column")
print(df["Drawingsheet"].max())
print("min value of textbook column")
print(df["Textbook"].min())
new_Totalunit=df['Total unit'] + 5
print(df['Total unit'])
```
